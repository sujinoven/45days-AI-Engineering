```python
#importing libraries
import os
import numpy as np
from openai import OpenAI
from dotenv import load_dotenv
```

```python
#Loading api key

load_dotenv()

client=OpenAI(api_key=os.getenv("OPENAI_API_KEY"))
```

```python
#Generating the first embedding

text="Artificial Intelligence helps computers learn."

response=client.embeddings.create(
    model="text-embedding-3-small",
    input=text)

embedding=response.data[0].embedding

print(embedding)
```

```python
print("Number of dimensions:", len(embedding))

#The default output for text-embedding-3-small has 1,536 dimensions.Each dimension is a number in the vector. Together, they represent the text in a form that can be compared mathematically.
```

```python
#Generate embeddings for multiple documents

documents = [
    "Python is a popular programming language.",
    "Machine learning allows computers to learn from data.",
    "RAG combines document retrieval with language generation.",
    "Neural networks are inspired by the human brain."
]
```

```python
#Creating a function
def get_embedding(text):

    response = client.embeddings.create(
        model="text-embedding-3-small",
        input=text
    )

    return response.data[0].embedding
```

```python
#generate vectors for every document
document_embeddings = []

for document in documents:

    embedding = get_embedding(document)

    document_embeddings.append(embedding)

print("Documents:", len(documents))
print("Embeddings:", len(document_embeddings))
```

```python
#Cosine similarity

def cosine_similarity(vector_a, vector_b):

    vector_a = np.array(vector_a)
    vector_b = np.array(vector_b)

    dot_product = np.dot(vector_a, vector_b)

    norm_a = np.linalg.norm(vector_a)
    norm_b = np.linalg.norm(vector_b)

    similarity = dot_product / (norm_a * norm_b)

    return similarity
```

```python
#Semantic Search
query = "How can AI retrieve information from documents?"

query_embedding = get_embedding(query)

results = []

for i, document_embedding in enumerate(document_embeddings):

    score = cosine_similarity(
        query_embedding,
        document_embedding
    )

    results.append((score, documents[i]))

results.sort(key=lambda item: item[0], reverse=True)

for score, document in results:

    print("Similarity:", round(float(score), 4))
    print("Document:", document)
    print()
```
