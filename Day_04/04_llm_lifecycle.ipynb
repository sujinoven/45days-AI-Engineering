```python
#Defining tiny vocabulary and transition table

next_token_probabilities = {
    "artificial": {
        "intelligence": 0.8,
        "technology": 0.2
    },
    "intelligence": {
        "is": 1.0
    },
    "is": {
        "powerful": 0.7,
        "useful": 0.3
    },
    "powerful": {
        "today": 1.0
    },
    "useful": {
        "today": 1.0
    }
}
```

```python
#Next token prediction function

import random

def predict_next_token(current_token):
    options = next_token_probabilities.get(current_token)

    if options is None:
        return None

    tokens = list(options.keys())
    probabilities = list(options.values())

    return random.choices(
        tokens,
        weights=probabilities
    )[0]
```

```python
#Generating text one token at a time
def generate_text(start_token, max_tokens=5):
    result = [start_token]

    current_token = start_token

    for _ in range(max_tokens):
        next_token = predict_next_token(current_token)

        if next_token is None:
            break

        result.append(next_token)
        current_token = next_token

    return " ".join(result)


print(generate_text("artificial"))
```
