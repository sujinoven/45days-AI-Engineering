```python
#Defining possible next tokens

import random

next_token_probabilities = {
    "Artificial": {
        "intelligence": 1.0
    },
    "intelligence": {
        "can": 1.0
    },
    "can": {
        "learn": 0.6,
        "help": 0.25,
        "reason": 0.15
    },
    "learn": {
        "from": 1.0
    },
    "help": {
        "people": 1.0
    },
    "reason": {
        "about": 1.0
    },
    "from": {
        "data": 1.0
    },
    "people": {
        "solve": 1.0
    },
    "about": {
        "problems": 1.0
    },
    "data": {
        "today": 1.0
    },
    "solve": {
        "problems": 1.0
    },
    "problems": {
        "today": 1.0
    },
    "today": {
        "<END>": 1.0
    }
}
```

```python
#Creating next token prediction function

def predict_next_token(current_token):

    possibilities = next_token_probabilities[current_token]

    tokens = list(possibilities.keys())
    probabilities = list(possibilities.values())

    next_token = random.choices(
        tokens,
        weights=probabilities,
        k=1
    )[0]

    return next_token
```

```python
#Generating the complete response

def generate_text(start_token, max_tokens=10):

    generated_tokens = [start_token]

    current_token = start_token

    for _ in range(max_tokens):

        next_token = predict_next_token(current_token)

        if next_token == "<END>":
            break

        generated_tokens.append(next_token)

        current_token = next_token

    return " ".join(generated_tokens)
```

```python
print(generate_text("Artificial"))
```
