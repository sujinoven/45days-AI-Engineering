```python
#Importign libraries and connecting open ai

from dotenv import load_dotenv
from openai import OpenAI
import os

load_dotenv()

api_key=os.getenv("OPENAI_API_KEY")

if not api_key:
    raise ValueError("API Key not found. Check your .env file.")

client=OpenAI(api_key=api_key)

print("Successfully connected to OpenAI!")
```

```python
#Creating reusable function

def ask_ai(instructions,user_prompt):

    response=client.responses.create(
        model="gpt-4o-mini",
        instructions=instructions,
        input=user_prompt
    )

    return response.output_text
```

```python
#Testing a weak prompt

instructions="You are a helpful assistant."

user_prompt="Explain APIs"

answer=ask_ai(instructions,user_prompt)

print(answer)
```

```python
#Structured prompt

instructions = """
You are an experienced programming instructor.

Teach -+
programming concepts to beginners.
Use simple English.
Include practical examples.
Avoid unnecessary technical jargon.
"""

user_prompt = """
Explain APIs.

Organize your response into:
1. Definition
2. How an API works
3. A real-world example
4. A simple Python example
"""

answer = ask_ai(instructions, user_prompt)

print(answer)
```

```python
#Few shot prompting
instructions = """
You are a sentiment classification assistant.

Classify each sentence as Positive, Negative, or Neutral.
Return only the sentiment label.
"""

user_prompt = """
Example 1:
Text: I love this product.
Label: Positive

Example 2:
Text: This is terrible.
Label: Negative

Example 3:
Text: It works as expected.
Label: Neutral

Now classify:
Text: The product is okay, but I expected more.
Label:
"""

answer = ask_ai(instructions, user_prompt)

print(answer)
```
