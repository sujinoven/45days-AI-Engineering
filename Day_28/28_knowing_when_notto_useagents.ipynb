## Approach 3: AI Agent
An agent can decide which tools to use and what steps to take next.

Example: research a topic.

It might:
1. Search the web.
2. Read relevant documents.
3. Identify missing information.
4. Search again.
5. Summarize its findings.
Best when the path is not fully predefined
An agent can adapt its actions based on observations. That flexibility also introduces more complexity.

# Building Today's Decision System

```python
#Defining Example tasks
tasks = [
    {
        "name": "Calculate monthly expenses",
        "needs_language_understanding": False,
        "needs_multiple_dynamic_steps": False,
        "needs_external_tools": False
    },
    {
        "name": "Summarize a customer review",
        "needs_language_understanding": True,
        "needs_multiple_dynamic_steps": False,
        "needs_external_tools": False
    },
    {
        "name": "Research a company and compare competitors",
        "needs_language_understanding": True,
        "needs_multiple_dynamic_steps": True,
        "needs_external_tools": True
    }
]
```

```python
#Creating the decision function
def choose_approach(task):

    if task["needs_multiple_dynamic_steps"] and task["needs_external_tools"]:
        return "AI Agent"

    elif task["needs_language_understanding"]:
        return "LLM Workflow"

    else:
        return "Python Workflow"
```

```python
#Testing the decision system
for task in tasks:

    approach = choose_approach(task)

    print("Task:", task["name"])
    print("Recommended approach:", approach)
    print("-" * 50)
```

```python
#Simple Cost Comparison
approaches = {
    "Python Workflow": {
        "model_calls": 0,
        "complexity": "Low",
        "predictability": "High"
    },
    "LLM Workflow": {
        "model_calls": 1,
        "complexity": "Medium",
        "predictability": "Medium"
    },
    "AI Agent": {
        "model_calls": 4,
        "complexity": "High",
        "predictability": "Variable"
    }
}

for name, details in approaches.items():
    print(name)
    print("Model calls:", details["model_calls"])
    print("Complexity:", details["complexity"])
    print("Predictability:", details["predictability"])
    print("-" * 40)
```
