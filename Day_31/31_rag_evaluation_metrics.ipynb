```python
#Creating a test dataset
test_cases = [
    {
        "question": "What is RAG?",
        "relevant_ids": ["D1", "D4", "D5"],
        "retrieved_ids": ["D1", "D4", "D2"]
    },
    {
        "question": "What are embeddings?",
        "relevant_ids": ["D2", "D3"],
        "retrieved_ids": ["D2", "D3"]
    },
    {
        "question": "What is a vector database?",
        "relevant_ids": ["D4", "D6"],
        "retrieved_ids": ["D4", "D1", "D2"]
    }
]
```

```python
#Calculating Precision
def calculate_precision(relevant_ids, retrieved_ids):

    if len(retrieved_ids) == 0:
        return 0.0

    relevant_retrieved = set(relevant_ids) & set(retrieved_ids)

    precision = len(relevant_retrieved) / len(retrieved_ids)

    return precision
```

```python
#Calculating Recall
def calculate_recall(relevant_ids, retrieved_ids):

    if len(relevant_ids) == 0:
        return 0.0

    relevant_retrieved = set(relevant_ids) & set(retrieved_ids)

    recall = len(relevant_retrieved) / len(relevant_ids)

    return recall
```

```python
#Calculate F1
def calculate_f1(precision, recall):

    if precision + recall == 0:
        return 0.0

    f1 = 2 * (precision * recall) / (precision + recall)

    return f1
```

```python
#Evaluating all test cases
evaluation_results = []

for test in test_cases:

    precision = calculate_precision(
        test["relevant_ids"],
        test["retrieved_ids"]
    )

    recall = calculate_recall(
        test["relevant_ids"],
        test["retrieved_ids"]
    )

    f1 = calculate_f1(precision, recall)

    result = {
        "question": test["question"],
        "precision": round(precision, 3),
        "recall": round(recall, 3),
        "f1": round(f1, 3)
    }

    evaluation_results.append(result)

for result in evaluation_results:
    print(result)
```

```python
#Calculating Average metrics
average_precision = sum(
    result["precision"] for result in evaluation_results
) / len(evaluation_results)

average_recall = sum(
    result["recall"] for result in evaluation_results
) / len(evaluation_results)

average_f1 = sum(
    result["f1"] for result in evaluation_results
) / len(evaluation_results)

print("Average Precision:", round(average_precision, 3))
print("Average Recall:", round(average_recall, 3))
print("Average F1:", round(average_f1, 3))
```
