```python
import json

#Creating a JSON string

json_text="""{
"name":"SUJIN",
"email":"sujinoven02@gmail.com",
"company":"Microsoft"
}"""

print(json_text)
```

```python
#convert json text into python dictionary

data=json.loads(json_text)

print(data)
print(type(data))
```

```python
#accessing individual values

print("name:",data["name"])
print("Email:",data["email"])
print("Company:",data["company"])
```

```python
#Converts a Python idctionary back into JSON

json_output=json.dumps(data,indent=4)

print(json_output)
```

# mini-project: Contact Information Extractor

```python
from openai import OpenAI
from dotenv import load_dotenv
import os
import json

load_dotenv()

client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))
```

```python
#Craete the extractor
def extract_contact(text):
    response = client.responses.create(
        model="gpt-4.1-mini",

        instructions="""
        Extract the person's name, email, and company.
        Return null when information is missing.
        Never invent information.
        """,

        input=text,

        text={
            "format": {
                "type": "json_schema",
                "name": "contact_extraction",
                "strict": True,
                "schema": contact_schema
            }
        }
    )

    return json.loads(response.output_text)
```

```python
#Defining Schema
contact_schema = {
    "type": "object",
    "properties": {
        "name": {
            "type": ["string", "null"]
        },
        "email": {
            "type": ["string", "null"]
        },
        "company": {
            "type": ["string", "null"]
        }
    },
    "required": ["name", "email", "company"],
    "additionalProperties": False
}
```

```python
email_text = """
Hello,

My name is David Lee.
I am a software developer at Amazon.
My email address is david.lee@example.com.

Regards,
David
"""

result = extract_contact(email_text)

print(result)
```

```python
#Testing with missing information

email_text_2 = """
Hello,

I am Priya.
I am interested in learning Python.

Thanks.
"""

result_2 = extract_contact(email_text_2)

print(result_2)
```
