```python

# Day 02 - Training a Single Neuron
# Goal: Learn the pattern y = 2x + 1

# 1. Training examples
training_data = [
    (0, 1),
    (1, 3),
    (2, 5),
    (3, 7),
    (4, 9),
]

# 2. Initial parameters
weight = 0.0
bias = 0.0
learning_rate = 0.01
epochs = 1000

# 3. Training loop
for epoch in range(epochs):
    total_loss = 0.0
    weight_gradient = 0.0
    bias_gradient = 0.0

    for x, target in training_data:
        # Forward pass
        prediction = weight * x + bias

        # Calculate squared error
        error = prediction - target
        total_loss += error ** 2

        # Calculate gradients
        weight_gradient += 2 * error * x
        bias_gradient += 2 * error

    n = len(training_data)

    # Average gradients
    weight_gradient /= n
    bias_gradient /= n

    # Update parameters
    weight -= learning_rate * weight_gradient
    bias -= learning_rate * bias_gradient

    # Display progress
    if epoch % 100 == 0:
        print(
            f"Epoch {epoch}: "
            f"Loss = {total_loss / n:.6f}"
        )

# 4. Learned parameters
print("\nTraining complete!")
print(f"Learned weight: {weight:.4f}")
print(f"Learned bias: {bias:.4f}")

# 5. Test on an unseen input
test_x = 10
prediction = weight * test_x + bias

print(f"\nInput: {test_x}")
print(f"Prediction: {prediction:.4f}")
print("Expected answer: 21")
```
