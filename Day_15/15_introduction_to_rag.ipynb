```python
#Creating documents
documents = [
    "Python is a programming language.",
    "Machine learning allows computers to learn from data.",
    "RAG combines document retrieval with language generation."
]
```

```python
#BAsic Search function

def search_documents(query,document):
    results=[]

    query_words=query.lower().split()

    for document in documents:
        score=0

        for word in query_words:
            if word in document.lower():
                score +=1

        results.append((score,document))

    results.sort(reverse=True)

    return results
```

```python
#Testing it

query="What is RAG?"

results=search_documents(query,documents)

for score,document in results:
    print("Score:",score)
    print("Document:",document)
    print()
```

```python
#Removing ? and checking

import string

def search_documents(query, documents):

    results = []

    # Remove punctuation and lowercase the query
    query = query.lower().translate(
        str.maketrans("", "", string.punctuation)
    )

    query_words = query.split()

    for document in documents:

        score = 0

        # Clean the document
        clean_document = document.lower().translate(
            str.maketrans("", "", string.punctuation)
        )

        document_words = clean_document.split()

        for word in query_words:

            if word in document_words:
                score += 1

        results.append((score, document))

    results.sort(reverse=True)

    return results
```

```python
query = "What is RAG?"

results = search_documents(query, documents)

for score, document in results:

    print("Score:", score)
    print("Document:", document)
    print()
```

# The two documents scoring 1 are tied because each matches one query word. Our current sorting also uses the document text to break ties.
