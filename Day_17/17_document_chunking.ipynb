```python
document = """
Python is a popular programming language.

Python is widely used in artificial intelligence.

Machine learning is a branch of artificial intelligence.

Machine learning algorithms learn patterns from data.

RAG stands for Retrieval-Augmented Generation.

RAG retrieves relevant information from documents
before generating an answer.
"""
```

```python
print(document)
```

```python
#1.Fixed Size chunking

def fixed_size_chunking(text, chunk_size):

    chunks = []

    for i in range(0, len(text), chunk_size):

        chunk = text[i:i + chunk_size]

        chunks.append(chunk)

    return chunks
```

```python
chunks = fixed_size_chunking(document, 100)

for i, chunk in enumerate(chunks):

    print(f"CHUNK {i + 1}")
    print(chunk)
    print("----------------")
```

```python
#Strategy : 2 Chunking with overlap

def chunk_with_overlap(text, chunk_size, overlap):

    chunks = []

    start = 0

    while start < len(text):

        end = start + chunk_size

        chunk = text[start:end]

        chunks.append(chunk)

        start += chunk_size - overlap

    return chunks
```

```python
overlap_chunks = chunk_with_overlap(
    document,
    chunk_size=100,
    overlap=20
)

for i, chunk in enumerate(overlap_chunks):

    print(f"CHUNK {i + 1}")
    print(chunk)
    print("----------------")
```

```python
#Startegy 3: Sentence based chunking

def sentence_chunking(text):

    sentences = text.split(".")

    chunks = []

    for sentence in sentences:

        sentence = sentence.strip()

        if sentence:

            chunks.append(sentence + ".")

    return chunks
```

```python
sentence_chunks = sentence_chunking(document)

for i, chunk in enumerate(sentence_chunks):

    print(f"CHUNK {i + 1}: {chunk}")
```
