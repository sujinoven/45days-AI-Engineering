## Today's objective: Build an interactive Python program that simulates the basic stages of an LLM.

By the end, you should understand:
1. How text is split into tokens.
2. How tokens can be represented as IDs.
3. How next-token prediction works.
4. How temperature affects token selection.
5. How a generated sequence becomes a response.

```python
#Building a simple tokenizer


def tokenize(text):
    text=text.lower()
    tokens=text.split()
    return tokens

sentence="Artificial Intelligence is powerful"

tokens=tokenize(sentence)

print("Original Text:",sentence)
print("Tokens:",tokens)
print("Number of Tokens:",len(tokens))
```

```python
#Converting tokens into IDs
vocabulary= {
    "artificial":1,
    "intelligence":2,
    "is":3,
    "powerful":4,
    "machine":5,
    "learning":6
}

token_ids=[vocabulary.get(token,0) for token in tokens]

print("Tokens:",tokens)
print("Token IDs:",token_ids)
```

```python
import random

next_token_probabilities = {
    "artificial": {
        "intelligence": 1.0
    },
    "intelligence": {
        "is": 1.0
    },
    "is": {
        "powerful": 0.6,
        "useful": 0.4
    },
    "powerful": {
        "today": 1.0
    },
    "useful": {
        "today": 1.0
    },
    "today": {
        "<END>": 1.0
    }
}

def predict_next_token(current_token):

    possibilities = next_token_probabilities[current_token]

    possible_tokens = list(possibilities.keys())
    probabilities = list(possibilities.values())

    next_token = random.choices(
        possible_tokens,
        weights=probabilities,
        k=1
    )[0]

    return next_token
```

```python
print(predict_next_token("is"))
```

```python
#Generating a Complete response
```

```python
def generate_response(start_token):

    generated_tokens = [start_token]

    current_token = start_token

    while True:

        next_token = predict_next_token(current_token)

        if next_token == "<END>":
            break

        generated_tokens.append(next_token)

        current_token = next_token

    return " ".join(generated_tokens)


response = generate_response("artificial")

print("Generated response:")
print(response)
```

```python
#Temperature experiment

def apply_temperature(probabilities, temperature):

    if temperature <= 0:
        raise ValueError("Temperature must be greater than zero.")

    adjusted = [
        probability ** (1 / temperature)
        for probability in probabilities
    ]

    total = sum(adjusted)

    normalized = [
        value / total
        for value in adjusted
    ]

    return normalized


original_probabilities = [0.6, 0.4]

for temperature in [0.5, 1.0, 2.0]:

    result = apply_temperature(
        original_probabilities,
        temperature
    )

    print("Temperature:", temperature)
    print("Adjusted probabilities:", result)
    print()
```
