```python
from openai import OpenAI
import os
from dotenv import load_dotenv
import json

load_dotenv()

from openai import OpenAI
from openai import (
    APIConnectionError,
    APIStatusError,
    AuthenticationError,
    BadRequestError,
    RateLimitError
)

import time

client=OpenAI(api_key=os.getenv("OPENAI_API_KEY"))
```

```python
#Error handling
try:
    response = client.responses.create(
        model="gpt-4.1-mini",
        input="Explain artificial intelligence in simple words."
    )

    print(response.output_text)

except Exception as error:
    print("Something went wrong!")
    print(error)
```

```python
#Handling special API errors
try:
    response = client.responses.create(
        model="gpt-4.1-mini",
        input="Explain what an API is."
    )

    print(response.output_text)

except AuthenticationError:
    print("Authentication failed. Check your API key.")

except RateLimitError:
    print("Rate limit reached. Please wait before trying again.")

except APIConnectionError:
    print("Connection problem. Check your internet connection.")

except BadRequestError as error:
    print("Your request is invalid:", error)

except APIStatusError as error:
    print("API returned an error:", error.status_code)

except Exception as error:
    print("Unexpected error:", error)
```

```python
#Retry function
def ask_ai_with_retry(prompt, max_retries=3):

    for attempt in range(max_retries + 1):

        try:
            response = client.responses.create(
                model="gpt-4.1-mini",
                input=prompt
            )

            return response.output_text

        except (RateLimitError, APIConnectionError) as error:

            if attempt == max_retries:
                print("Maximum retries reached.")
                raise

            wait_time = 2 ** attempt

            print(f"Temporary error: {error}")
            print(f"Retrying in {wait_time} seconds...")

            time.sleep(wait_time)

        except (AuthenticationError, BadRequestError):
            print("This error needs to be fixed. Not retrying.")
            raise
```

## Today's mini-project
Build a Reliable AI Assistant.

Your assistant should:
- Accept a question from the user.
- Send it to the OpenAI API.
- Retry temporary failures.
- Display the answer when successful.
- Display a useful message if the request ultimately fails.## 

```python
def reliable_assistant():

    question = input("Ask your AI assistant: ")

    try:
        answer = ask_ai_with_retry(question)

        print("\nAI Assistant:")
        print(answer)

    except Exception as error:
        print("\nSorry, the request could not be completed.")
        print("Error:", error)


reliable_assistant()
```
